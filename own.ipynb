```python
#tokenization
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
#stemming
from nltk.stem import WordNetLemmatizer
#lemmatization
from nltk.stem import PorterStemmer
```

```python
sentences = [
    "i am learning python and python",
    "we are happy today",
    "i am sad today"
]

def preprocess_text(documents):

    english_stopwords = stopwords.words("english")
    import string
    punctuations = string.punctuation

    stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()

    all_final_tokens = []   # To store results for all documents

    for doc in documents:

        # 1. Lowercase
        raw_text = doc.lower()
        print("\nAfter lowercase:", raw_text)

        # 2. Tokenization
        tokens = word_tokenize(raw_text)
        print("After tokenization:", tokens)

        # 3. Stopword removal
        filtered_tokens = []
        for t in tokens:
            if t not in english_stopwords:
                filtered_tokens.append(t)
        print("After stopword removal:", filtered_tokens)

        # 4. Remove punctuation + keep only alphabetic words
        clean_tokens = []
        for t in filtered_tokens:
            if t.isalpha():        # removes punctuation and numbers
                clean_tokens.append(t)
        print("After cleaning:", clean_tokens)

        # 5. Lemmatization
        final_tokens = []
        for t in clean_tokens:
            final_tokens.append(lemmatizer.lemmatize(t))
        print("After lemmatization:", final_tokens)

        # Store result of this document
        all_final_tokens.append(final_tokens)

    return all_final_tokens


cleaned_sentences = preprocess_text(sentences)
print("Cleaned Sentences :",cleaned_sentences)
```
