```python
#NLP phase-1: Text Processing
#Text->cleaning->tokenization->stopwords removal->stemming/lementization
```

```python
text="I am learning python programming and Python is fast!!.."
print("Original text: ",text)
```

```python
#1-convert to lower case
text=text.lower()
print("LowerCase :",text)
```

```python
#2-remove puntuations
#re- regex,s-spaces,w-word,r-raw
import re
text=re.sub(r'[^\w\s]','',text)
print("Without punctuation :",text)
#remove everything except letters and spaces
```

```python
#3-tokenization
tokens=text.split()
print("Tokens :",tokens)
```

```python
#4-remove stopwords
stopwords=["a","is","an","the","and","i"]
filtered_tokens=[]
for word in tokens:
  if word not in stopwords:
    filtered_tokens.append(word)
print("Filtered Tokens after removing stopwords :",filtered_tokens)
```

```python
#5-stemming
def stem(word):
  if word.endswith("ing"):
    return word[:-3]
  return word
stemmed_tokens=[]
for word in filtered_tokens:
  stemmed_tokens.append(stem(word))
print("Stemmed Tokens :",stemmed_tokens)
```

```python
#will convert text->count
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sentences=[
    "i am learning python and python",
    "we are happy today",
    "i am sad todey"
]
```

```python
vectorizer=CountVectorizer()
X=vectorizer.fit_transform(sentences)
#learns vocabulary
#converts sentences into numbers
```

```python
print("Vocabulary :",vectorizer.get_feature_names_out())
#all unique words
# sorted in alphabetical order
#remove single letters like i,a
```

```python
print("Vectors:")
print(X.toarray())
# Each row is a sentence
# Each column is a word
# each value is a count
```

```python
#TF-IDF = Term Frequency -Inverse Document
#why TF-IDF
#because count is not enough
# bag of words treats all words equally
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
sentences=[
    "i am learning python and python",
    "we are happy today",
    "i am sad todey"
]
```

```python
vectorizer= TfidfVectorizer()
X=vectorizer.fit_transform(sentences)
# builds vocabulary
#calculate TF
# Calculate idf -impact of each word
#multiples -tf x idf
# produces matrix

```

```python
print("TF-IDF Matrix :")
print(X.toarray())
```

```python
# Count+Impact of each word
#TF-IDF rewards rare important words and penalizes common words
```

```python
# TF =count(words)/ total words in sentences
# IDF = log(total number of sentences/ number of sentences with th word)
```
